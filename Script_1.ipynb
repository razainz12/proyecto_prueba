```python
import pandas as pd
```

```python
df = pd.read_excel("dataset/Pokemon.xlsx")
df.head(5)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Análisis Descriptivo
print("=" * 80)
print("ANÁLISIS DESCRIPTIVO DEL DATASET DE POKÉMON")
print("=" * 80)

# 1. Información general
print("\n1. INFORMACIÓN GENERAL:")
print(f"   - Dimensiones: {df.shape[0]} filas, {df.shape[1]} columnas")
print(f"   - Nombres de columnas: {list(df.columns)}")

# 2. Tipos de datos
print("\n2. TIPOS DE DATOS:")
print(df.dtypes)

# 3. Valores nulos
print("\n3. VALORES NULOS:")
print(df.isnull().sum())
print(f"   - Porcentaje de valores faltantes: {(df.isnull().sum().sum() / (df.shape[0] * df.shape[1]) * 100):.2f}%")

# 4. Estadísticas descriptivas
print("\n4. ESTADÍSTICAS DESCRIPTIVAS:")
print(df.describe())

# 5. Información de columnas categóricas
print("\n5. COLUMNAS CATEGÓRICAS:")
categorical_cols = df.select_dtypes(include='object').columns
for col in categorical_cols:
    print(f"\n   {col}:")
    print(f"      - Valores únicos: {df[col].nunique()}")
    print(f"      - Top 5: {df[col].value_counts().head().to_dict()}")

# 6. Correlación (para variables numéricas)
print("\n6. CORRELACIÓN (Variables Numéricas):")
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
if len(numeric_cols) > 1:
    print(df[numeric_cols].corr().to_string())
else:
    print("   No hay suficientes variables numéricas para calcular correlación")

print("\n" + "=" * 80)
```

```python
# Visualizaciones del Análisis Descriptivo
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Gráfico 1: Distribución de variables numéricas
numeric_data = df.select_dtypes(include=['int64', 'float64'])
if not numeric_data.empty and len(numeric_data.columns) > 0:
    numeric_data.iloc[:, 0].hist(ax=axes[0, 0], bins=30, edgecolor='black')
    axes[0, 0].set_title('Distribución de Primera Variable Numérica')
    axes[0, 0].set_xlabel('Valor')
    axes[0, 0].set_ylabel('Frecuencia')

# Gráfico 2: Valores nulos
df.isnull().sum().plot(kind='bar', ax=axes[0, 1], color='coral')
axes[0, 1].set_title('Cantidad de Valores Nulos por Columna')
axes[0, 1].set_xlabel('Columnas')
axes[0, 1].set_ylabel('Nulos')
axes[0, 1].tick_params(axis='x', rotation=45)

# Gráfico 3: Matriz de correlación
if len(numeric_data.columns) > 1:
    sns.heatmap(numeric_data.corr(), annot=True, fmt='.2f', ax=axes[1, 0], cmap='coolwarm')
    axes[1, 0].set_title('Matriz de Correlación')
else:
    axes[1, 0].text(0.5, 0.5, 'Pocas variables numéricas', ha='center', va='center')

# Gráfico 4: Información del dataset
axes[1, 1].axis('off')
info_text = f"""
RESUMEN DEL DATASET
━━━━━━━━━━━━━━━━━━━━━━━━━
Filas: {df.shape[0]}
Columnas: {df.shape[1]}
Memoria: ~{df.memory_usage(deep=True).sum() / 1024:.2f} KB
Tipo de datos:
  • Numéricos: {len(numeric_data.columns)}
  • Categóricos: {len(df.select_dtypes(include='object').columns)}
Nulos totales: {df.isnull().sum().sum()}
"""
axes[1, 1].text(0.1, 0.5, info_text, fontfamily='monospace', fontsize=10, verticalalignment='center')

plt.tight_layout()
plt.show()

print("✓ Análisis descriptivo completado")
```
