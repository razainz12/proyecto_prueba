# Análisis Gráfico Descriptivo - Dataset Pokémon
## Visualización y exploración de datos mediante gráficos

Este notebook genera visualizaciones descriptivas detalladas del dataset de Pokémon usando pandas, matplotlib y seaborn.

## 1. Importar librerías

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import warnings
import os
warnings.filterwarnings('ignore')

# Configurar estilos
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 10

# Crear carpeta para guardar gráficos
output_dir = 'graficos'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

print("✓ Librerías importadas correctamente")
print(f"✓ Carpeta de salida: {os.path.abspath(output_dir)}")
```

## 2. Cargar y revisar el dataset

```python
# Cargar dataset
df = pd.read_excel("dataset/Pokemon.xlsx")

print(f"Dataset cargado: {df.shape[0]} filas x {df.shape[1]} columnas\n")
print("Primeras filas del dataset:")
print(df.head())
print("\nInformación del dataset:")
print(df.info())
print("\nResumen de valores nulos:")
print(df.isnull().sum())
```

## 3. Preparar datos para visualización

```python
# Separar variables numéricas y categóricas
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df.select_dtypes(include='object').columns.tolist()

print(f"Variables numéricas ({len(numeric_cols)}): {numeric_cols}")
print(f"\nVariables categóricas ({len(categorical_cols)}): {categorical_cols}")

# Estadísticas de variables numéricas
print("\nEstadísticas de variables numéricas:")
print(df[numeric_cols].describe())
```

## 4. Gráficos de distribución numérica

```python
# Distribución de características principales
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Distribución de Características Principales de Pokémon', fontsize=16, fontweight='bold')

features = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8', '#F7DC6F']

for idx, (ax, feature, color) in enumerate(zip(axes.flat, features, colors)):
    ax.hist(df[feature], bins=30, color=color, edgecolor='black', alpha=0.7)
    ax.set_title(f'Distribución de {feature}', fontweight='bold')
    ax.set_xlabel(f'{feature}')
    ax.set_ylabel('Frecuencia')
    ax.grid(axis='y', alpha=0.3)
    
    # Añadir media y mediana
    mean_val = df[feature].mean()
    median_val = df[feature].median()
    ax.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Media: {mean_val:.1f}')
    ax.axvline(median_val, color='green', linestyle='--', linewidth=2, label=f'Mediana: {median_val:.1f}')
    ax.legend()

plt.tight_layout()
plt.savefig(os.path.join(output_dir, '01_distribucion_caracteristicas.png'), dpi=300, bbox_inches='tight')
plt.show()

print("✓ Gráficos de distribución completados y guardados")
```

```python
# Gráficos de caja (Box plots)
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Análisis de Distribución: Diagramas de Caja', fontsize=16, fontweight='bold')

for idx, (ax, feature) in enumerate(zip(axes.flat, features)):
    sns.boxplot(data=df, y=feature, ax=ax, color='#87CEEB', palette='Set2')
    ax.set_title(f'Box Plot de {feature}', fontweight='bold')
    ax.grid(axis='y', alpha=0.3)
    
    # Mostrar valores atípicos
    q1 = df[feature].quantile(0.25)
    q3 = df[feature].quantile(0.75)
    iqr = q3 - q1
    outliers = df[(df[feature] < q1 - 1.5*iqr) | (df[feature] > q3 + 1.5*iqr)][feature]
    ax.text(0.05, 0.95, f'Atípicos: {len(outliers)}', transform=ax.transAxes, 
            verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig(os.path.join(output_dir, '02_box_plots.png'), dpi=300, bbox_inches='tight')
plt.show()

print("✓ Diagramas de caja completados y guardados")
```

```python
# Gráficos de densidad (Density plots)
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Gráficos de Densidad', fontsize=16, fontweight='bold')

for idx, (ax, feature, color) in enumerate(zip(axes.flat, features, colors)):
    ax.clear()
    df[feature].plot(kind='density', ax=ax, color=color, linewidth=2.5, legend=False)
    ax.fill_between(range(len(df)), df[feature].min(), df[feature].max(), alpha=0.1, color=color)
    ax.set_title(f'Densidad de {feature}', fontweight='bold')
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(output_dir, '03_density_plots.png'), dpi=300, bbox_inches='tight')
plt.show()

print("✓ Gráficos de densidad completados y guardados")
```

## 5. Gráficos de frecuencia categórica

```python
# Gráfico 1: Tipos de Pokémon (Type 1)
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Distribución de Tipos de Pokémon', fontsize=16, fontweight='bold')

# Count plot para Type 1
type1_counts = df['Type 1'].value_counts()
sns.barplot(x=type1_counts.index, y=type1_counts.values, ax=axes[0], palette='husl')
axes[0].set_title('Pokémon por Tipo Principal (Type 1)', fontweight='bold')
axes[0].set_xlabel('Tipo')
axes[0].set_ylabel('Cantidad')
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(axis='y', alpha=0.3)

# Añadir valores en las barras
for i, (idx, v) in enumerate(type1_counts.items()):
    axes[0].text(i, v + 1, str(v), ha='center', va='bottom')

# Count plot para Type 2
type2_counts = df['Type 2'].value_counts()
sns.barplot(x=type2_counts.index, y=type2_counts.values, ax=axes[1], palette='Set2')
axes[1].set_title('Pokémon por Tipo Secundario (Type 2)', fontweight='bold')
axes[1].set_xlabel('Tipo')
axes[1].set_ylabel('Cantidad')
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(axis='y', alpha=0.3)

# Añadir valores en las barras
for i, (idx, v) in enumerate(type2_counts.items()):
    axes[1].text(i, v + 1, str(v), ha='center', va='bottom')

plt.tight_layout()
plt.savefig(os.path.join(output_dir, '04_distribucion_tipos.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"✓ Total de tipos únicos (Type 1): {df['Type 1'].nunique()}")
print(f"✓ Total de tipos secundarios (Type 2): {df['Type 2'].nunique()}")
print("✓ Gráficos de tipos guardados")
```

```python
# Gráfico 2: Generación de Pokémon
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Análisis por Generación', fontsize=16, fontweight='bold')

# Barplot de Generación
gen_counts = df['Generation'].value_counts().sort_index()
sns.barplot(x=gen_counts.index, y=gen_counts.values, ax=axes[0], palette='viridis')
axes[0].set_title('Pokémon por Generación', fontweight='bold')
axes[0].set_xlabel('Generación')
axes[0].set_ylabel('Cantidad')
axes[0].grid(axis='y', alpha=0.3)

# Añadir valores en las barras
for i, (gen, count) in enumerate(gen_counts.items()):
    axes[0].text(i, count + 1, str(count), ha='center', va='bottom', fontweight='bold')

# Pie chart de Pokémon Legendarios vs No Legendarios
legendary_counts = df['Legendary'].value_counts()
colors_pie = ['#FF6B6B', '#4ECDC4']
axes[1].pie(legendary_counts.values, labels=['No Legendario', 'Legendario'], 
            autopct='%1.1f%%', colors=colors_pie, startangle=90, 
            textprops={'fontsize': 11, 'fontweight': 'bold'})
axes[1].set_title('Distribución: Pokémon Legendarios vs No Legendarios', fontweight='bold')

plt.tight_layout()
plt.savefig(os.path.join(output_dir, '05_generacion_legendarios.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"✓ Legendarios: {legendary_counts.get(True, 0)}")
print(f"✓ No Legendarios: {legendary_counts.get(False, 0)}")
print("✓ Gráficos de generación guardados")
```

## 6. Matriz de correlación y scatter plots

```python
# Matriz de correlación
fig, ax = plt.subplots(figsize=(12, 10))

# Calcular correlación
correlation_matrix = df[numeric_cols].corr()

# Crear heatmap
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=0.5, cbar_kws={"shrink": 0.8}, ax=ax)
ax.set_title('Matriz de Correlación entre Variables Numéricas', fontsize=14, fontweight='bold', pad=20)

plt.tight_layout()
plt.savefig(os.path.join(output_dir, '06_matriz_correlacion.png'), dpi=300, bbox_inches='tight')
plt.show()

print("✓ Matriz de correlación completada y guardada")
```

```python
# Scatter plots de variables relevantes
fig, axes = plt.subplots(2, 2, figsize=(14, 12))
fig.suptitle('Relaciones entre Variables Clave', fontsize=16, fontweight='bold')

# Plot 1: Attack vs Defense
scatter1 = axes[0, 0].scatter(df['Attack'], df['Defense'], 
                               c=df['Total'], cmap='viridis', s=100, alpha=0.6, edgecolors='black')
axes[0, 0].set_xlabel('Attack', fontweight='bold')
axes[0, 0].set_ylabel('Defense', fontweight='bold')
axes[0, 0].set_title('Attack vs Defense (coloreado por Total)', fontweight='bold')
axes[0, 0].grid(alpha=0.3)
plt.colorbar(scatter1, ax=axes[0, 0], label='Total')

# Plot 2: Sp. Atk vs Sp. Def
scatter2 = axes[0, 1].scatter(df['Sp. Atk'], df['Sp. Def'], 
                               c=df['Generation'], cmap='plasma', s=100, alpha=0.6, edgecolors='black')
axes[0, 1].set_xlabel('Sp. Atk', fontweight='bold')
axes[0, 1].set_ylabel('Sp. Def', fontweight='bold')
axes[0, 1].set_title('Sp. Atk vs Sp. Def (coloreado por Generación)', fontweight='bold')
axes[0, 1].grid(alpha=0.3)
plt.colorbar(scatter2, ax=axes[0, 1], label='Generation')

# Plot 3: Speed vs Total
scatter3 = axes[1, 0].scatter(df['Speed'], df['Total'], 
                               c=df['HP'], cmap='Spectral', s=100, alpha=0.6, edgecolors='black')
axes[1, 0].set_xlabel('Speed', fontweight='bold')
axes[1, 0].set_ylabel('Total', fontweight='bold')
axes[1, 0].set_title('Speed vs Total Stats (coloreado por HP)', fontweight='bold')
axes[1, 0].grid(alpha=0.3)
plt.colorbar(scatter3, ax=axes[1, 0], label='HP')

# Plot 4: HP vs Attack (por Tipo Legendario)
for legendary in [False, True]:
    mask = df['Legendary'] == legendary
    label = 'Legendario' if legendary else 'No Legendario'
    axes[1, 1].scatter(df[mask]['HP'], df[mask]['Attack'], 
                      label=label, s=100, alpha=0.6, edgecolors='black')
axes[1, 1].set_xlabel('HP', fontweight='bold')
axes[1, 1].set_ylabel('Attack', fontweight='bold')
axes[1, 1].set_title('HP vs Attack (por Tipo Legendario)', fontweight='bold')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(output_dir, '07_scatter_plots.png'), dpi=300, bbox_inches='tight')
plt.show()

print("✓ Scatter plots completados y guardados")
```

```python
# Pair plot (relaciones entre múltiples variables)
print("Generando pair plot (esto puede tomar unos segundos)...")
selected_cols = ['HP', 'Attack', 'Defense', 'Speed', 'Total']
pair_plot = sns.pairplot(df[selected_cols], diag_kind='hist', 
                         plot_kws={'alpha': 0.6, 'edgecolor': 'k', 's': 50},
                         diag_kws={'bins': 20, 'edgecolor': 'k'})
pair_plot.fig.suptitle('Pair Plot de Variables Principales', fontsize=16, fontweight='bold', y=1.001)
plt.tight_layout()
plt.savefig(os.path.join(output_dir, '08_pair_plot.png'), dpi=300, bbox_inches='tight')
plt.show()

print("✓ Pair plot completado y guardado")
```
